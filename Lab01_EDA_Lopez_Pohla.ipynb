# 7.5. Frecuencias de una categoría


```python
print(columnas_categoricas)
# Cambia el indice si la primera columna no es adecuada.
columna_categorica = columnas_categoricas[0]
print("Columna elegida:", columna_categorica)
```

```python
frecuencia = df[columna_categorica].value_counts(
  dropna=False
)
porcentaje = df[columna_categorica].value_counts(
  dropna=False,
  normalize=True
).mul(100).round(2)
tabla_frecuencias = pd.DataFrame({
  "cantidad": frecuencia,
  "porcentaje": porcentaje
})
tabla_frecuencias.head(15)
```

La categoría más frecuente es Movie, con aproximadamente 69,62 % de los registros. TV Show representa aproximadamente 30,38 %. Por lo tanto, las categorías no están equilibradas, ya que las películas representan más del doble de los registros que las series. En este caso no se observan formas diferentes de escritura para estas dos categorías.

# 8.2. Histograma con media y mediana


```python
media = df[columna_numerica].mean()
mediana = df[columna_numerica].median()
plt.figure(figsize=(10, 5))
sns.histplot(
  data=df,
  x=columna_numerica,
  bins=20,
  kde=True,
  color="#0068B3"
)
plt.axvline(
  media,
  color="#C0392B",
  linestyle="--",
  label=f"Media: {media:.2f}"
)
plt.axvline(
  mediana,
  color="#007A6A",
  linestyle="-.",
  label=f"Mediana: {mediana:.2f}"
)
plt.title(f"Distribucion de {columna_numerica}")
plt.xlabel(columna_numerica)
plt.ylabel("Cantidad de registros")
plt.legend()
plt.show()
```

La mayoría de los valores de release_year se concentra en los años más recientes, especialmente desde aproximadamente 2010 en adelante. La distribución presenta una cola hacia los años más antiguos, porque existen títulos con fechas de lanzamiento bastante anteriores. La media y la mediana están relativamente separadas, con una media cercana a 2014 y una mediana de 2017. También existen valores alejados correspondientes a títulos antiguos. La forma de la distribución depende de que el dataset reúne contenido de distintas épocas y no solamente lanzamientos recientes.

# 8.3. Diagrama de caja


```python
plt.figure(figsize=(10, 3.5))
sns.boxplot(
  x=df[columna_numerica],
  color="#8ED1C6"
)
plt.title(f"Diagrama de caja de {columna_numerica}")
plt.xlabel(columna_numerica)
plt.show()
```

El diagrama de caja muestra que existen observaciones alejadas hacia los años más antiguos. Estos valores pueden considerarse poco frecuentes dentro de la distribución, pero no necesariamente son errores, pueden corresponder a películas o series antiguas que forman parte del catálogo de Netflix. Por eso, no sería correcto eliminarlos automáticamente sin revisar primero el significado de esos registros.

# 8.4. Gráfico de barras para una columna categórica


```python
top_categorias = df[columna_categorica].value_counts(
  dropna=False
).head(10)
plt.figure(figsize=(10, 5))
sns.barplot(
  x=top_categorias.values,
  y=top_categorias.index.astype(str),
  color="#009C88"
)
plt.title(f"Categorias mas frecuentes de {columna_categorica}")
plt.xlabel("Cantidad de registros")
plt.ylabel(columna_categorica)
plt.tight_layout()
plt.show()
```

La categoría Movie domina claramente el gráfico, mientras que TV Show tiene una cantidad bastante menor de registros. Por lo tanto, los grupos no están equilibrados. El gráfico muestra solamente las categorías principales, aunque en este caso type tiene solo esas dos categorías principales. El resultado coincide con la tabla de frecuencias, donde las películas representan aproximadamente el 69,62 % y las series el 30,38 %.
